## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
Classification (Binary). We are predicting a definitive, mutually exclusive outcome: whether a customer will churn (1) or stay (0) within the next 30 days. This is chosen over scoring or ranking because the immediate business action is binary—either an automated retention playbook is triggered, or it is not.$0
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
The target variable is **customer churn within the next 30 days**. A customer is labeled as **1 (churned)** if they cancel their subscription, stop purchasing, or become inactive according to the business definition during the observation window. Otherwise, they are labeled **0 (retained)**. This label comes from an **observed outcome** recorded in historical customer data rather than a manually assigned rule. Using observed outcomes allows the model to learn real behavioral patterns associated with customer attrition.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
The primary success metric is **F1 Score**. Customer churn datasets are often imbalanced because most customers stay while only a smaller percentage leave. F1 Score balances Precision and Recall, ensuring that the model identifies likely churners without generating excessive false alarms. A defensible target would be an **F1 Score above 0.75**, indicating useful performance for retention decision-making.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
One row represents **one customer at a specific observation date**. Each row contains customer attributes such as usage behavior, engagement metrics, account information, transaction history, support interactions, and the churn label. The dataframe structure allows the model to learn relationships between customer behavior patterns and future churn outcomes.
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
Customer churn is influenced by multiple interacting factors including usage frequency, spending behavior, support requests, engagement trends, tenure, and historical activity. These relationships are often nonlinear and difficult to express with simple if-statements. For example, a small drop in activity may be normal for one customer but a strong churn signal for another depending on their history. Machine learning can discover these complex patterns automatically and provide more accurate decision support than fixed business rules.
```
